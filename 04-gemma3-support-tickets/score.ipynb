# Lesson 4: sort support tickets with Gemma 3 270M

Training happens in **Unsloth Studio** (no code). This notebook only keeps score: the base model before, the Studio model after, on the same 1,170 test tickets.

```python
from mlx_lm import load
from tickets import QUEUES, load_tickets, latest_studio_run, ask, score

tickets = load_tickets("test")
print(len(tickets), "test tickets")
print(QUEUES)
```

```python
tickets[0]
```

## Before: the base model

```python
model, tokenizer = load("unsloth/gemma-3-270m-it")

for t in tickets[:4]:
    print(f"{ask(model, tokenizer, t['text'])!r:<16} true: {t['label']:<13} {t['text']}")
```

```python
before = score(model, tokenizer, tickets)
```

## Train in Unsloth Studio

Model `unsloth/gemma-3-270m-it`, dataset `cngchis/Support-Ticket-Router-12K-Cleaned`, text → User, label → Assistant, 300 steps, batch 8, learning rate 1e-4 (`studio-config.yaml`). Then come back here.

## After: the model trained in Studio

```python
run = latest_studio_run()
print(run)

model, tokenizer = load("unsloth/gemma-3-270m-it", adapter_path=run)

for t in tickets[:4]:
    print(f"{ask(model, tokenizer, t['text'], instruction=False)!r:<16} true: {t['label']:<13} {t['text']}")
```

```python
after = score(model, tokenizer, tickets, instruction=False)
```

```python
print(f"before {100 * before:.1f} %  →  after {100 * after:.1f} %")
```
